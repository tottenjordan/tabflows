# Complete AutoML Tabular Workflows Feature Showcase

This notebook presents a comprehensive showcase of advanced features available in `tabflows` for AutoML Tabular workflows on Vertex AI Pipelines.

Key features highlighted in this showcase include:
- **Stratified Data Splitting**: Splitting datasets according to target label distribution (`stratified_split_key="deposit"`).
- **Sample Weighting**: Incorporating observation weights during model training (`weight_column="sample_weight"`).
- **Custom Ops Export Configuration**: Exporting additional TF/SavedModel artifacts without custom TensorFlow operations (`export_additional_model_without_custom_ops=True`).
- **Stage 1 Full Search Pipeline**: Complete hyperparameter search & architecture exploration.
- **Stage 2 Fast Retraining (Skip Evaluation)**: Fast model retraining pipeline bypassing the evaluation stage for rapid iteration.
- **Stage 2 Distillation Retraining (Skip Evaluation)**: Knowledge distillation pipeline combined with skip-evaluation for optimized student model training.
- **Artifact Inspection & Summary**: Verifying exported configuration parameters and pipeline template artifacts.

```python
from dotenv import load_dotenv

from tabflows import (
    TabularPipelineConfig,
    create_distill_skip_evaluation_pipeline_job,
    create_skip_evaluation_pipeline_job,
    create_tabular_pipeline_job,
)

# Load environment variables from local .env file if available
load_dotenv()
print("Libraries and imports loaded successfully.")
```

```python
# Initialize TabularPipelineConfig with stratified splitting,
# sample weighting, and export without custom ops
config = TabularPipelineConfig(
    project_id="ci-showcase-project",
    bucket_uri="gs://ci-showcase-bucket",
    stratified_split_key="deposit",
    weight_column="sample_weight",
    export_additional_model_without_custom_ops=True,
    tuning_result_output="gs://ci-showcase-bucket/automl_tabular_pipeline/tuning_result",
)

print(f"Project ID: {config.project_id}")
print(f"Bucket URI: {config.bucket_uri}")
print(f"Stratified Split Key: {config.stratified_split_key}")
print(f"Weight Column: {config.weight_column}")
print(
    "Export Additional Model Without Custom Ops: "
    f"{config.export_additional_model_without_custom_ops}"
)
```

```python
# Compile Stage 1 Full Search pipeline with stratified splitting and sample weighting
stage_1_job = create_tabular_pipeline_job(
    config=config,
    job_id="showcase-full-search",
    log_experiment=False,
)

print(f"Stage 1 Pipeline Job created: {type(stage_1_job).__name__}")
print(f"Stage 1 Display Name: {stage_1_job.job_id}")
print(f"Pipeline Root Dir: {config.root_dir}")
```

```python
# Compile Stage 2 Fast Skip-Evaluation retraining pipeline
skip_eval_job = create_skip_evaluation_pipeline_job(
    config=config,
    job_id="showcase-skip-evaluation",
    log_experiment=False,
)

print(f"Stage 2 Fast Skip-Evaluation Pipeline Job created: {type(skip_eval_job).__name__}")
print(f"Skip-Evaluation Display Name: {skip_eval_job.job_id}")
```

```python
# Compile Stage 2 Distilled Skip-Evaluation pipeline
distill_skip_eval_job = create_distill_skip_evaluation_pipeline_job(
    config=config,
    job_id="showcase-distill-skip-evaluation",
    log_experiment=False,
)

print(
    "Stage 2 Distilled Skip-Evaluation Pipeline Job created: "
    f"{type(distill_skip_eval_job).__name__}"
)
print(f"Distill Skip-Evaluation Display Name: {distill_skip_eval_job.job_id}")
```

```python
# Inspect export model artifacts and configuration summary
print("=== AutoML Tabular Workflows Feature Showcase Summary ===")
print(f"Target Column: {config.target_column}")
print(f"Prediction Type: {config.prediction_type}")
print(f"Optimization Objective: {config.optimization_objective}")
print(f"Stratified Split Key: {config.stratified_split_key}")
print(f"Weight Column: {config.weight_column}")
print(
    "Export Additional Model Without Custom Ops: "
    f"{config.export_additional_model_without_custom_ops}"
)
print(f"Tuning Result Output URI: {config.tuning_result_output}")
print(f"Stage 1 Job Template: {stage_1_job.job_id}")
print(f"Stage 2 Fast Job Template: {skip_eval_job.job_id}")
print(f"Stage 2 Distill Job Template: {distill_skip_eval_job.job_id}")
print("\nAll pipeline templates compiled and feature showcase inspection complete.")
```
