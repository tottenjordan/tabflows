# Specialized Optimization Objectives & Precision/Recall Target Tuning

This notebook demonstrates how to configure and compile AutoML Tabular pipelines with specialized optimization objectives and target threshold constraints in Vertex AI:

1. **Fraud Detection Objective**: Optimize for `maximize-precision-at-recall` with a target recall constraint (e.g. 0.95) to minimize false positive alerts while capturing 95% of fraudulent transactions.
2. **Marketing Conversion Objective**: Optimize for `maximize-recall-at-precision` with a target precision constraint (e.g. 0.90) to maximize outreach campaign volume while ensuring high lead quality (90% precision).
3. **Experiment Tracking**: Compile and log specialized objective pipeline jobs into Vertex AI Experiments for comparative analysis.

```python

from dotenv import load_dotenv

from tabflows import (
    TabularPipelineConfig,
    create_tabular_pipeline_job,
    list_experiment_runs,
)

# Load environment variables from local .env file
load_dotenv()
print("Environment setup and tabflows imports completed.")
```

```python
# Configuration for Fraud Detection objective (Maximize Precision at 95% Recall)
fraud_config = TabularPipelineConfig(
    optimization_objective="maximize-precision-at-recall",
    optimization_objective_recall_value=0.95,
    train_budget_milli_node_hours=1000,
    experiment_name="specialized-objectives-experiment",
)

print(f"Fraud Detection Config Project ID: {fraud_config.project_id}")
print(f"Optimization Objective: {fraud_config.optimization_objective}")
print(f"Target Recall Constraint: {fraud_config.optimization_objective_recall_value}")
print(f"Training Budget (milli node hours): {fraud_config.train_budget_milli_node_hours}")
```

```python
# Compile Fraud Detection AutoML Tabular PipelineJob with 0.95 recall constraint
fraud_pipeline_job = create_tabular_pipeline_job(
    config=fraud_config,
    job_id="fraud-detection-max-precision-re095",
    log_experiment=True,
)

print("Fraud Detection PipelineJob compiled successfully:")
print(f"Job ID / Display Name: {fraud_pipeline_job.job_id}")
print(f"Pipeline Root: {fraud_config.root_dir}")
```

```python
# Configuration for Marketing Conversion objective (Maximize Recall at 90% Precision)
marketing_config = TabularPipelineConfig(
    optimization_objective="maximize-recall-at-precision",
    optimization_objective_precision_value=0.90,
    train_budget_milli_node_hours=1000,
    experiment_name="specialized-objectives-experiment",
)

print(f"Marketing Conversion Config Project ID: {marketing_config.project_id}")
print(f"Optimization Objective: {marketing_config.optimization_objective}")
print(f"Target Precision Constraint: {marketing_config.optimization_objective_precision_value}")
print(f"Training Budget (milli node hours): {marketing_config.train_budget_milli_node_hours}")
```

```python
# Compile Marketing Conversion AutoML Tabular PipelineJob with 0.90 precision constraint
marketing_pipeline_job = create_tabular_pipeline_job(
    config=marketing_config,
    job_id="marketing-conversion-max-recall-pr090",
    log_experiment=True,
)

print("Marketing Conversion PipelineJob compiled successfully:")
print(f"Job ID / Display Name: {marketing_pipeline_job.job_id}")
print(f"Pipeline Root: {marketing_config.root_dir}")
```

```python
# Query and list logged Vertex AI Experiment runs comparing the specialized objectives
experiment_runs = list_experiment_runs(
    experiment_name="specialized-objectives-experiment",
    config=fraud_config,
)

if experiment_runs is not None:
    print(f"Logged experiment runs for 'specialized-objectives-experiment': {len(experiment_runs)}")
    print(experiment_runs)
else:
    print("No experiment runs found or experiment history empty in mock environment.")
```
